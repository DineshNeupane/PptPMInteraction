```python
import dask
dask.config.set({"dataframe.convert-string": False})
import dask.dataframe as dd
```

```python
PM_DESIGNATION = "pm10"
```

```python
pm_metro_dask = dd.read_csv(r'\data\PM_PLUS_METRO_2014_2023_HHMMSS.csv', assume_missing=True)
pm_metro_df = pm_metro_dask.compute()
```

```python
pm_metro_df.head()
```

```python
pm_metro_df[PM_DESIGNATION].isna().sum()
```

```python
pm_metro_df = pm_metro_df.dropna(subset=[PM_DESIGNATION])
```
