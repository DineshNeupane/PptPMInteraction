```python
import matplotlib as mpl
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

mpl.rcParams.update(mpl.rcParamsDefault)

pm10 = pd.read_csv(r"\results\pm10\PM10_removal_by_baseline_pm_median_reference.csv")
pm25 = pd.read_csv(r"\results\pm25\PM25_removal_by_baseline_pm_median_reference.csv")

PM25_DARK, PM10_DARK = "#D55E00", "#0072B2"
PM25_LIGHT50, PM10_LIGHT50 = "#EEBF99", "#99C7E0"

pm_labels = [
    "<5", "7.5", "10", "12.5", "15",
    "17.5", "20", "22.5", "25",
    "27.5", "30", "32.5", "35", "37.5", "40+"
]

fig, ax = plt.subplots(figsize=(6, 5))

plot_info = [
    (pm25, r"PM$_{2.5}$", PM25_DARK, PM25_LIGHT50),
    (pm10, r"PM$_{10}$", PM10_DARK, PM10_LIGHT50),
]

for dfr, label, line_color, fill_color in plot_info:
    x = np.arange(len(dfr))

    ax.plot(
        x,
        dfr["theta"],
        marker="o",
        linewidth=2,
        color=line_color,
        markerfacecolor=line_color,
        markeredgecolor=line_color,
        label=label
    )

    ax.fill_between(
        x,
        dfr["theta_ci_low"],
        dfr["theta_ci_high"],
        color=fill_color,
        alpha=0.5
    )

ax.axhline(0, linestyle="--", linewidth=1, color="black")

ax.set_xlabel("Baseline PM Bin (µg/m³)")
ax.set_ylabel(r"Est. removal rate (% mm$^{-1}$),  $\hat{\theta}$")

ax.set_xticks(np.arange(len(pm_labels)))
ax.set_xticklabels(pm_labels, rotation=45, ha="right")

ax.legend()

ax.grid(True, alpha=0.3, axis="y")

plt.tight_layout()

plt.savefig(
    r"\figures\removal_by_baseline_theta.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

pm_labels = [
    "<5", "7.5", "10", "12.5", "15",
    "17.5", "20", "22.5", "25",
    "27.5", "30", "32.5", "35", "37.5", "40+"
]

fig, ax = plt.subplots(figsize=(6, 5))

plot_info = [
    (pm25, r"PM$_{2.5}$", PM25_DARK, PM25_LIGHT50),
    (pm10, r"PM$_{10}$", PM10_DARK, PM10_LIGHT50),
]

for dfr, label, line_color, fill_color in plot_info:
    x = np.arange(len(dfr))

    ax.plot(
        x,
        dfr["theta"],
        marker="o",
        linewidth=2,
        color=line_color,
        markerfacecolor=line_color,
        markeredgecolor=line_color,
        label=label
    )

    ax.fill_between(
        x,
        dfr["theta_ci_low"],
        dfr["theta_ci_high"],
        color=fill_color,
        alpha=0.5
    )

ax.axhline(0, linestyle="--", linewidth=1, color="black")

ax.set_xlabel("Baseline PM Bin (µg/m³)")
ax.set_ylabel(r"Est. removal rate (% mm$^{-1}$)")

ax.set_xticks(np.arange(len(pm_labels)))
ax.set_xticklabels(pm_labels, rotation=45, ha="right")

ax.legend()

ax.grid(True, alpha=0.3, axis="y")

plt.tight_layout()

plt.savefig(
    r"\figures\removal_by_baseline_theta.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
