```python
import pandas as pd
import os
```

```python
folder_path = r"\data\aqs_data_10_years"
years = range(2014, 2024)
```

```python
csv_files = [os.path.join(folder_path, f"{year}.csv") for year in years]
```

```python
df_list = [pd.read_csv(file) for file in csv_files if os.path.exists(file)]
combined_df = pd.concat(df_list, ignore_index=True)
```

```python
combined_df.head()
```

```python
output_path = os.path.join(folder_path, "combined_2014_to_2023.csv")
combined_df.to_csv(output_path, index=False)
```

#### Assign UID to each lat/lng, only Keep first POC data

```python
coordinates_df = pd.read_csv(r'\data\coordinates\yearly_coordinates_pm25\unique_coordinates_with_years_and_id_pm25.csv')
pm_df = pd.read_csv(r'\data\aqs_data_10_years\combined_2014_to_2023.csv')
```

```python
merged = pd.merge(pm_df, coordinates_df, on=['Latitude', 'Longitude'], how='left')
```

```python
merged.head()
```

```python
df_sorted = merged.sort_values(by=['UID', 'Date GMT', 'Time GMT'])
```

```python
df_deduped = df_sorted.drop_duplicates(subset=['UID', 'Date GMT', 'Time GMT'], keep='first')
```

```python
df_deduped.reset_index(drop=True, inplace=True)
```

```python
df_deduped.head()
```

```python
df_deduped.to_csv(r'\data\aqs_data_10_years\combined_2014_to_2023_UID_no_POC.csv', index=False)
```
