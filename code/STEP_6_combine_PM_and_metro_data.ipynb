```python
import pandas as pd
import dask
dask.config.set({"dataframe.convert-string": False})
import dask.dataframe as dd
```

```python
metro_df_dask = dd.read_csv(r'\data\all_parallel_processed_data_2014_2023.csv', assume_missing=True)
metro_df = metro_df_dask.compute()
metro_df.head()
```

```python
metro_df.shape
```

```python
pm_df_dask = dd.read_csv(r'\data\combined_2014_to_2023_UID_no_POC.csv')
pm_df = pm_df_dask.compute()
pm_df.head()
```

```python
pm_df = pm_df.drop(columns=['Years_Present', 'Datum', 'POC'])
pm_df.head()
```

```python
pm_df = pm_df.dropna(subset=['UID'])
```

```python
pm_df['date_time'] = pd.to_datetime(pm_df['Date GMT'] + ' ' + pm_df['Time GMT'])
pm_df.head()
```

```python
pm_df['date_time'] = pd.to_datetime(pm_df['date_time'])
print(pm_df['date_time'].dtype)
```

```python
metro_df['date_time'] = pd.to_datetime(metro_df['date_time'], format='%Y-%m-%d %H:%M:%S', errors='coerce')
```

```python
merged_df = metro_df.merge(
    pm_df[['UID', 'date_time', 'pm25']],
    on=['UID', 'date_time'],
    how='left'  # or 'inner' if you only want matching rows
)
```
