```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import dask
dask.config.set({"dataframe.convert-string": False})
```

```python
df = pd.read_csv(r"\data\aqs_data_10_years\10_years_metro_pm10_data.csv", dtype={"UID": "int64"})
df.describe()
```

```python
df_smoke = pd.read_csv(r"\data\smoke\Smoke_PM10_10_years.csv")
df_smoke.describe()
```

```python
df['DateTime'] = pd.to_datetime(df['date_time'])
```

```python
df_smoke['DateTime'] = pd.to_datetime(df_smoke['DateTime'])
```

```python
df = df.drop(['date_time'], axis=1)
```

```python
merged_df = df.merge(
    df_smoke[["DateTime", "UID", "Density"]],
    on=["DateTime", "UID"],
    how="left"
)
```

```python
merged_df.head()
```

```python
merged_df["Density"] = merged_df["Density"].fillna("NS")
```

```python
merged_df.to_csv(r"\data\aqs_data_10_years\10_years_metro_pm10_smoke_data.csv")
```
